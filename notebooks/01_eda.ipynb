# 01 - Exploratory Data Analysis
This notebook performs EDA on the ACIS insurance dataset.

## Setup and Data Loading
Import the core libraries, load the insurance dataset, and create derived metrics for loss ratio and margin.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

DATA_PATH = Path('..') / 'data' / 'insurance_data.csv'
df = pd.read_csv(DATA_PATH, sep='\t', parse_dates=['TransactionDate'])
# derived metrics
df['LossRatio'] = df['TotalClaims'] / df['TotalPremium']
df['Margin'] = df['TotalPremium'] - df['TotalClaims']
df.head()
```

## Data Summary
Generate basic descriptive statistics to verify data types and inspect the first view of the dataset.

```python
# Basic summaries
df.describe(include='all')
```

## Provincial Loss Ratio
Create a visualization that compares average loss ratio across provinces to identify geographic risk patterns.

```python
# Plot LossRatio by Province
import seaborn as sns
plt.figure(figsize=(8,5))
sns.barplot(data=df, x='Province', y='LossRatio')
plt.title('Average Loss Ratio by Province')
plt.tight_layout()
plt.show()
```
